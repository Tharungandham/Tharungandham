```python
# Program Functionalities:
# //				Add a new contact
# //
# //				Display all contacts
# //
# //				Search for a contact by name
# //
# //				Remove a contact by name
# //
# //				Check if a contact with a given email exists
# //
# //				Count total contacts
def contact():
  con={}
  fn=True
  while fn:
    name=input("enter the name:")
    if name==".":
      break
    pn=input("enter the pn:")
    email=input("enter the email:")
    ls=[pn,email]
    con[name]=ls

  # Display all contacts after input is done
  print("\n--- All Contacts ---")
  for name, details in con.items():
      print(f"Name: {name}, Phone: {details[0]}, Email: {details[1]}")
  print("--------------------")

  src=input("enter a name to find:")
  # Use get() to safely retrieve the value or None if the key doesn't exist
  found_contact = con.get(src)
  if found_contact:
      print("found name:", found_contact)
  else:
      print(f"Contact '{src}' not found.")


  rmv=input("enter a name to remove:")
  # Use pop() with a default value to safely remove a key
  removed_contact = con.pop(rmv, None)
  if removed_contact:
      print("removed name:", removed_contact)
      print("--- Updated Contacts ---")
      for name, details in con.items():
          print(f"Name: {name}, Phone: {details[0]}, Email: {details[1]}")
      print("--------------------")
  else:
      print(f"Contact '{rmv}' not found for removal.")

  # This part is for checking if an email exists, not removing
  emai_to_find = input("enter an email to find:")
  email_found = False
  for name, details in con.items():
      if details[1] == emai_to_find:
          print(f"Email '{emai_to_find}' found for contact: {name}")
          email_found = True
          break # Found the email, no need to check further
  if not email_found:
      print(f"Email '{emai_to_find}' not found in contacts.")


  count = len(con) # Use len() to get the number of items in the dictionary
  print("\nTotal number of contacts remaining:", count)
  # print("Remaining contacts dictionary:", con) # Optional: print the dictionary itself

```

Define a Member class.

Each member has a name, member ID, and a list of borrowed books.

Add the following functionality:

A method to display member info.

A method to borrow a book (adds to the list).

A method to return a book (removes from the list).

A class method that keeps track of the total number of members registered.

A static method that checks if a book is eligible to be borrowed (e.g., books like "dictionary" are not borrowable).

```python
class Member:
  total_mem=0
  def __init__(self,name,id):
    self.mname=name
    self.mid=id
    self.mborrowed_books=[]
    Member.total_mem= Member.total_mem+1
  def display_mem_info(self):
    print("member name :",self.mname)
    print("member id :",self.mid)
  def borrow_book(self):
    borrowed=True
    while(borrowed):
      book=input("enter a book you want to borrow :")
      if book==".":
        borrowed=False
        break
      else:
        self.mborrowed_books.append(book)
  def return_book(self):
    book_rtn=input("enter the book to return :")
    if  book_rtn in self.mborrowed_books:
      self.mborrowed_books.remove(book_rtn)
      print(self.mborrowed_books)
    else:
      print("this book is not registered in your name")
  @classmethod
  def show_total_mem(cls):
    print("total members registered today in library",cls.total_mem)
  @staticmethod
  def eligibility(book):
    if book.endswith("dictionary"):
      print("dictionary is not for borrowing")
    else:
      print("you can borrow this book")

if __name__=="__main__":
  mb1=Member("tharun",18)
  mb1.display_mem_info()
  Member.eligibility("english_dictionary")
  mb1.borrow_book()
  mb1.return_book()
  Member.show_total_mem()
```

Define a BankAccount class.

Each account has an account holder name, account number, and balance.

Include:

A method to deposit and withdraw money.

A method to check the balance.

A class method to show the interest rate for all accounts (assume one common rate).

A static method to validate account numbers (e.g., account numbers should be exactly 10 digits).



```python
class BankAccount:
  interest=0
  def __init__(self,name,number):
    self.acname=name
    self.__accnumber=number
    self.__acbalance=0
    BankAccount.interest+=1
  @staticmethod
  def valid_accnumber(accnumber):
    if isinstance(accnumber,int) and len(str(accnumber)) ==10 :
      print("vaild acc number ")
    else:
      print("invalid account number ")
  def deposit(self):
    money=int(input("enter the amount you want to deposit in your account :"))
    if money>0:
      self.__acbalance=self.__acbalance+money
    else:
      print("invalid amount ")
  def withdraw(self):
    money=int(input("enter the amount you want to withdraw from your account :"))
    if money>0:
      self.__acbalance=self.__acbalance-money
    else:
      print("invalid amount ")
  def check_balance(self,accnumber):
    if self.__accnumber==accnumber:
      print("acount balance :",self.__acbalance)
    else:
      print("enter a valid accnumber")
  @classmethod
  def interse_rate(cls):
    print(cls.interest)

if __name__=="__main__":
  ba1=BankAccount("tharun",9132388180)
  BankAccount.valid_accnumber(9132388180)
  ba1.deposit()
  ba1.withdraw()
  ba1.check_balance(9132388180)
  BankAccount.interse_rate()
```

```python
class car:
  no_of_cars=0
  def __init__(self,mdl,year,enumber):
    self.model=mdl
    self.year=year
    self.engnumber=enumber
    car.no_of_cars+=1
  @staticmethod
  def validation(year):
    if 1990<year<2025:
      print("valid year")
    else:
      print("invalid year")
  def display_details(self):
    print("car details")
    print("car model :",self.model)
    print("car year :",self.year)
    print("car engine number",self.engnumber)
  @classmethod
  def cars_manufactured(cls):
    print("no of cars manufatured",cls.no_of_cars)
if __name__=="__main__":
  cr1=car("volvo",2000,"g5ubj")
  car.validation(2000)
  cr1.display_details()
  cr2=car("toyoto",2020,"hgfgdhs")
  cr2.display_details()
  car.cars_manufactured()
```

Payment System (Inheritance + Method Overriding)
You're building a payment system that handles different payment modes: CreditCard, DebitCard, and UPI.

Requirements:
Create a base class Payment with a method make_payment(amount)

Each subclass should override make_payment() to show a different message:

e.g., "Processing credit card payment of $500"

Optionally: add a static method to validate amount (> 0)

```python
class Payment:
  def make_payment(self,amount):
    self.amt=amount
    print("amount paid successfully",self.amt,"$")
class Credit_card(Payment):
  def make_payment(self,amount):
    print("credit card paid successfully",amount,"$")
class Mobile_bill(Payment):
  def make_payment(self,amount):
    print("mobile bill paid successfully",amount,"$")
class Loan(Mobile_bill):
  def make_payment(self,amount=None):
    if amount is None:
        amount = int(input("Enter amount for loan payment: "))
    self.amount = amount
    print("Loan paid successfully:", self.amount, "$")
if __name__=="__main__":
  ln=Loan()
  pn=Payment()
  cc=Credit_card()
  mb=Mobile_bill()
  for i in [ln,pn,cc,mb]:
    i.make_payment(5000)
```

Create a loop that takes product prices from the user until they type "done".

Requirements:
Use try-except to catch:

ValueError if input can't be converted to a float

Sum up valid prices and print total when done

```python
#exception handling
import functools
class Handling:
  def __init__(self):
    self.lst=[]
  def total_price(self):
    true=True
    try:
      while(true):
        price=float(input("enter the product prices one by one :"))

        if price=="done":
          true=False
          break
        else:
          self.lst.append(price)
    except ValueError as e:
      print("exception reason:",e)
    except Exception as e:
      print(e)

    finally:
      total=functools.reduce(lambda x,y:x+y,self.lst)
      print(total)

h=Handling()
h.total_price()
```

```python
#bank withdrawl
class InsufficientFunds(Exception):
  pass
class BankWithdraw:
  def __init__(self):
    self.balance=0
  def deposit(self,damount):
    try:
      if damount>=0:
        self.balance+=damount
        print("your account credited with :",self.balance)
      else:
        raise ValueError("please deposite the correct amount")
    except Exception as e:
        print("inbuilt exception :",e)
  def withdraw(self):
    try:
      amount=float(input("enter the amount to withdraw :"))
      if amount<=self.balance:
        self.balance-=amount
        print("remaining balance :",self.balance)
      else:
        raise InsufficientFunds("insufficient funds in your account")
    except Exception as e:
        print("inbuilt exception :",e)
if __name__=="__main__":
  bw=BankWithdraw()
  bw.deposit(5000)
  bw.withdraw()


```

Task
Ask the user to input the name of the student to delete.

Read all lines from the file.

Remove the line that matches the name (hint: use if name.lower() not in line.lower()).

Write back the remaining lines to the file.



```python
class Student:
  def student_details(self,filename):
    with open(filename,"w") as f:
      for i in range(4):
        name=input("enter student name :")
        marks=input("enter student marks :")
        id=input("enter student id :")
        f.write(f"name:{name},marks:{marks},id:{id}\n")
  def read_student_details(self,filename):
    try:
      with open(filename,"r") as fr:
        print(fr.read())
    except FileNotFoundError as e:
      print(e)
  def student_delete(self,filename):
    std_delete=input("enter the student name to delete ")
    try:
      with open(filename,"r") as rd:
        lines=rd.readlines()
        with open(filename,"w") as rw:
          for line in lines:
            if  std_delete.lower() not in line.lower():
              rw.write(line)
            else:
              print(f"line deleted successfully {line}")

    except FileNotFoundError as e:
      print(e)

if __name__=="__main__":
  stdobj=Student()
  stdobj.student_details("student")
  stdobj.read_student_details("student")
  stdobj.student_delete("student")
  stdobj.read_student_details("student")
```

Create a class Student with attributes:

student_id (int)

name (str)

marks (float)

Create another class StudentManager with methods:

add_student() — Input student details, create a Student object, and store it in a list.

save_to_file(filename) — Save the list of students to a file using serialization (pickle).

load_from_file(filename) — Load student list from a file.

display_all() — Print all student details in a readable format.

search_student(student_id) — Search for a student by their ID and print their details.

delete_student(student_id) — Delete a student by ID from the list and update the file.

The program should provide a simple menu-driven interface to call these methods.



```python
import pickle
class Student:
  def __init__(self):
    for i in range(1):
      self.id=int(input("enter the student id : "))
      self.name=input("enter the student name : ")
      self.age=int(input("enter the student age : "))
      self.marks=float(input("enter the student marks : "))
class StudentManager:
  def add_student(self):
    s1=Student()
    self.stu_lst=[]
    self.stu_lst.append(s1)
  def save_to_file(self,filename):
    with open(filename,'wb') as wf:
      pickle.dump(self.stu_lst,wf)
  def load_file(self,filename):
    with open(filename,"rb") as fr:
      self.fileread=pickle.load(fr)
      print(self.fileread)
  def search_student(self):
    student_id=int(input("enter the student id to search :"))
    student_delete=int(input("enter the student id to delete :"))
    for stu in self.fileread:
      if stu.id==student_id:
        print(f"student id {stu.id},student name {stu.name},age {stu.age}, marks {stu.marks}")
      if stu.id==student_delete:
        self.fileread.remove(stu)
    print(self.fileread)


if __name__=="__main__":
  m=StudentManager()
  m.add_student()
  m.save_to_file("student")
  m.load_file("student")
  m.search_student()


```

```python
class Employee:
  def __init__(self,emp_id,emp_name,emp_salary,emp_department,emp_city):
    self.id=emp_id
    self.name=emp_name
    self.salary=emp_salary
    self.department=emp_department
    self.city=emp_city
class Employee_manager:
  def __init__(self):
    self.employee_lst=[]
  def add_emp(self,eid,ename,esalary,edep,ecity):
    emp=Employee(eid,ename,esalary,edep,ecity)
    self.employee_lst.append(emp)
  def emp_display(self):
    for employee in self.employee_lst:
      print(f"emp id {employee.id}, emp name {employee.name}, employee salary {employee.salary} ,employee dep {employee.department} ,employee city {employee.city}")
  def highest_sal(self):
    high_sal=max(emp.salary for emp in self.employee_lst)
    print("highest salary :",high_sal)

if __name__=="__main__":
  empm=Employee_manager()
  empm.add_emp(18,"tharun",150000,"datascience","hyderabad")
  empm.add_emp(25,"srujeeth",155000,"yogascience","hyderabad")
  empm.add_emp(101,"sridhar",145000,"datascience","hyderabad")
  empm.emp_display()
  empm.highest_sal()


```
